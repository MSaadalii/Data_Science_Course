## BroadCasting

```python
l = [1,2,3,4,5]
arr = np.array(l)
arr
```

```python
arr + 10
```

```python
slice = arr[:2]
slice
```

```python
slice = slice + 10
slice
```

```python
arr
```

```python
arr2 = arr
arr2
```

```python
arr2[0] = 44
arr2
```
