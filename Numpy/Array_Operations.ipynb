## Arithmetic Operations

```python
n1 = np.array([1,2,3,4,5])
n2 = np.array([6,7,8,9,10])
n1
```

```python
n1 + n2
```

```python
n1 - n2
```

```python
n1 * n2
```

```python
n1 / n2
```

```python
n1 // n2
```

```python
n1 ** n2
```

## BroadCasting

```python
l = [1,2,3,4,5]
arr = np.array(l)
arr
```

```python
arr + 10
```

```python
slice = arr[:2]
slice
```

```python
slice = slice + 10
slice
```

```python
arr
```

```python
arr2 = arr
arr2
```

```python
arr2[0] = 44
arr2
```

```python
arr
```

## Stacking of Arrays

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

## Spliting Arrays

```python
c = np.arange(16).reshape(4,4)
```

```python
np.hsplit(c,2)
```

```python
np.hsplit(c,4)
```

```python
np.vsplit(c,2)
```

```python
a = np.vsplit(c,4)
```

```python
for i in a:
    print(i)
```
