```python
import numpy as np
```

## **Slicing and Indexing of Matrix**

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr_1 = arr[0]
arr_1
```

```python
arr_1 = arr[3:7 ,3:6]
arr_1
```

```python
arr_col = arr[:,2]
arr_col
```

## **Boolean Indexing**

```python
array = np.arange(1,11)
array
```

```python
bool = array%2 == 0
bool
```

```python
array = array[bool]
array
```
